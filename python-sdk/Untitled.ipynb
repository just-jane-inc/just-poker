````python
import requests
import json
import openapi_client as poker_api
poker_url = 'http://localhost:7653'

def create_connection(base_url: str, token: str = '') -> poker_api.ApiClient:
    config = None
    if token:
      config = poker_api.Configuration(host=base_url, access_token=token)
    else:
      config = poker_api.Configuration(host=base_url)

    return poker_api.ApiClient(config)

def create_user(base_url: str, twitch_id: str, display_name: str) -> poker_api.UserApiKey | None:
    api = poker_api.UserApi(create_connection(base_url))
    dto = poker_api.UserUserDTO(display_name=display_name, twitch_id=twitch_id, user_type='bot')
    resp = api.user_post(poker_api.UserPostRequest(dto))
    return resp

def delete_user(base_url: str, api_key: str):
    api = poker_api.UserApi(create_connection(base_url, api_key))
    resp = api.user_me_delete()
    return resp
````

````python
user = create_user(poker_url, 'test-twitch-id', 'test-bot-meow')
````
